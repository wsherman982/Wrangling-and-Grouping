```python
# Step 4: Get unique Team values
csv_df["Team"].unique()

# Create the team mapping
team_mapping = {
    "Marketing": "Building1",
    "Client Services": "Building1",
    "Product": "Building1",
    "Sales": "Building1",
    "Finance": "Building2",
    "Legal": "Building2",
    "Human Resources": "Building2",
    "Engineering": "Building3",
    "Business Development": "Building3",
    "Distribution": "Building 4"
}

# Step 5: Create the location column
csv_df["location"] = csv_df["Team"].map(team_mapping)
csv_df["location"] = csv_df["location"].fillna("Administration")

csv_df[["Team", "location"]]
```
